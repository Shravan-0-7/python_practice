```python
#1 
orders = [
    {
        'id': 101,
        'customer': 'Asha',
        'status': 'DELIVERED',
        'items': [('keyboard', 2, 1499.0), ('mouse', 1, 599.0)]
    },
    {
        'id': 102,
        'customer': 'Ravi',
        'status': 'CANCELLED',
        'items': [('monitor', 1, 10999.0)]
    },
    {
        'id': 103,
        'customer': 'Asha',
        'status': 'DELIVERED',
        'items': [('cable', 5, 199.0)]
    },
    {
        'id': 104,
        'customer': 'Meera',
        'status': 'SHIPPED',
        'items': [('keyboard', 1, 1499.0), ('monitor', 1, 10999.0)]
    },
    {
        'id': 105,
        'customer': 'Ravi',
        'status': 'DELIVERED',
        'items': [('mouse', 3, 599.0)]
    }
]

```

```python
def analyse(orders):

    revenue = {}
    product_quantity = {}
    customer_status = {}
    for order in orders:
        customer = order['customer']
        status = order['status']
        # customer order status
        if customer not in customer_status:
            customer_status[customer] = True
        if status != 'DELIVERED':
            customer_status[customer] = False
        if status == 'CANCELLED':
            continue
        
        for item in order['items']:
            product, quantity, price = item
            amount = quantity * price
            revenue[customer] = revenue.get(customer, 0) + amount
            product_quantity[product] = (
                product_quantity.get(product, 0) + quantity
            )
    # B. Top Selling Products 
    sorted_products = sorted(
        product_quantity.items(),
        key=lambda x: (-x[1], x[0])
    )
    top_products = sorted_products[:2]
    # C. Fully Delivered Customers 
    fully_delivered = sorted(
        [customer for customer in customer_status
         if customer_status[customer]]
    )
    return {
        'revenue': revenue,
        'top_products': top_products,
        'fully_delivered': fully_delivered
    }
result = analyse(orders)
print(result)
```

```python
#Q2. Customer Data Cleaning

lines = [
    ' asha VERMA ; ASHA.V@Example.COM ; +91 98765-43210 ;pune ',
    'ravi kumar;ravi@@corp.in;9123456780;Mumbai',
    'meera;meera@mail;9123456780;Delhi',
    'john doe; john@x.org ;12345; Goa',
    'Neha Rao;neha.rao@firm.co.in;919876501234;nashik',
    'amit;amit@site.com;9876543210'
]
```

```python
def clean_customer(raw):
    fields = raw.split(';')
    if len(fields) != 4:
        return None
    name = fields[0].strip().title()
    email = fields[1].strip().lower()
    phone = fields[2].strip()
    city = fields[3].strip().title()
    # Validate name
    words = name.split()
    if len(words) == 0:
        return None
    for word in words:
        for ch in word:
            if not ch.isalpha():
                return None
    # Validate city
    if city == "":
        return None
    # Validate email
    if email.count('@') != 1:
        return None
    username, domain = email.split('@')
    if username == "":
        return None
    if '.' not in domain:
        return None
    if domain.startswith('.') or domain.endswith('.'):
        return None
    # Clean phone
    phone = phone.replace(" ", "")
    phone = phone.replace("-", "")
    if phone.startswith("+91"):
        phone = phone[3:]
    elif len(phone) == 12 and phone.startswith("91"):
        phone = phone[2:]
    # Validate phone
    if len(phone) != 10:
        return None
    for digit in phone:
        if not digit.isdigit():
            return None

    if phone[0] not in "6789":
        return None
    return {
        'name': name,
        'email': email,
        'phone': phone,
        'city': city
    }
# Process all records
rejected = 0
for raw in lines:
    result = clean_customer(raw)
    if result is not None:
        print(result)
    else:
        rejected += 1
print("Rejected:", rejected)
```
