```python
arr = [10, 34, 45, 6, 8, 20, 25, 19]
minimum = float('inf')
maximum = float('-inf')
prevMax = float('-inf')
prevMin = float('inf')

for i in arr:

    if i < minimum:
        prevMin = minimum
        minimum = i
    elif i < prevMin and i != minimum:
        prevMin = i

    if i > maximum:
        prevMax = maximum
        maximum = i
    elif i > prevMax and i != maximum:
        prevMax = i

print(f"Minimum number in arr is: {minimum}")
print(f"Maximum number in arr is: {maximum}")
print(f"Second minimum number is: {prevMin}")
print(f"Second maximum number is: {prevMax}")
```
